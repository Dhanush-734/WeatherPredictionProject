```python
from pathlib import Path
import pandas as pd
import joblib
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestRegressor

# Project root
ROOT = Path.cwd().parent

# Load featured dataset
df = pd.read_csv(ROOT / "dataset" / "weather_featured.csv")

# Features and Target
X = df.drop([c for c in ["Temperature", "Visibility"] if c in df.columns], axis=1)
y = df["Temperature"]

# Train-Test Split
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)

# Train model
rf = RandomForestRegressor(
    n_estimators=100,
    random_state=42
)

rf.fit(X_train, y_train)

# Save model
joblib.dump(
    rf,
    ROOT / "models" / "weather_model.pkl"
)

print("✅ Weather model trained and saved successfully!")
print("Saved to:", ROOT / "models" / "weather_model.pkl")
```
